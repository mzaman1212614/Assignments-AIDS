## Question 11: Student Performance Management System


```python

number_of_students = int(input("Enter number of students: "))

passed = 0
failed = 0
class_total_marks = 0
class_total_percentage = 0
highest_percentage = 0
lowest_percentage = 100

for student in range(number_of_students):

    print("\nStudent", student + 1)

    name = input("Enter student name: ")
    marks_in_python = int(input("Enter Python marks: "))
    marks_in_mathematics = int(input("Enter Mathematics marks: "))
    marks_in_english = int(input("Enter English marks: "))

    total_marks = (
        marks_in_python
        + marks_in_mathematics
        + marks_in_english
    )

    percentage = (total_marks / 300) * 100

    # Update class totals
    class_total_marks += total_marks
    class_total_percentage += percentage

    # Find highest and lowest percentage
    if percentage > highest_percentage:
        highest_percentage = percentage

    if percentage < lowest_percentage:
        lowest_percentage = percentage

    # Check pass or fail
    if (
        percentage >= 50
        and marks_in_python >= 40
        and marks_in_mathematics >= 40
        and marks_in_english >= 40
    ):
        result = "Pass"
        passed += 1

        # Assign grade only to passing students
        if percentage >= 90:
            grade = "A+"
        elif percentage >= 80:
            grade = "A"
        elif percentage >= 70:
            grade = "B"
        elif percentage >= 60:
            grade = "C"
        else:
            grade = "D"

    else:
        result = "Fail"
        grade = "N/A"
        failed += 1

    print("Name:", name)
    print("Total Marks:", total_marks, "/ 300")
    print(f"Percentage: {percentage:.2f}%")
    print("Result:", result)
    print("Grade:", grade)

# Calculate class average percentage
class_average = class_total_percentage / number_of_students

print("\n===== CLASS REPORT =====")
print("Total Students:", number_of_students)
print("Passed:", passed)
print("Failed:", failed)
print(f"Highest Percentage: {highest_percentage:.2f}%")
print(f"Lowest Percentage: {lowest_percentage:.2f}%")
print("Class Total Marks:", class_total_marks)
print(f"Class Average: {class_average:.2f}%")
```

## Question 12: Restaurant Ordering & Billing System

```python

total_bill = 0
total_items = 0
choice = 0

while choice != 6:

    print("\n========== MENU ==========")
    print("1. Burger Rs. 500")
    print("2. Pizza Rs. 1000")
    print("3. Biryani Rs. 350")
    print("4. Fries Rs. 250")
    print("5. Drink Rs. 150")
    print("6. Checkout")

    choice = int(input("Select an item (1-6): "))

    if choice == 1:
        total_bill += 500
        total_items += 1
        print("Burger added")

    elif choice == 2:
        total_bill += 1000
        total_items += 1
        print("Pizza added")

    elif choice == 3:
        total_bill += 350
        total_items += 1
        print("Biryani added")

    elif choice == 4:
        total_bill += 250
        total_items += 1
        print("Fries added")

    elif choice == 5:
        total_bill += 150
        total_items += 1
        print("Drink added")

    elif choice == 6:
        print("Checkout")

    else:
        print("Invalid choice. Please try again.")

# Calculate discount
if total_bill >= 5000:
    discount = total_bill * 0.20
elif total_bill >= 3000:
    discount = total_bill * 0.15
elif total_bill >= 1500:
    discount = total_bill * 0.10
else:
    discount = 0

# Calculate bill after discount
discounted_bill = total_bill - discount

# Calculate delivery charges
if discounted_bill >= 3000:
    delivery_charges = 0
else:
    delivery_charges = 200

# Calculate final bill
final_bill = discounted_bill + delivery_charges

# Display final bill
print("\n===== FINAL BILL =====")
print("Total Items:", total_items)
print("Original Bill: Rs.", total_bill)
print("Discount: Rs.", discount)
print("Delivery Charges: Rs.", delivery_charges)
print("Final Bill: Rs.", final_bill)
print("Thank you for ordering!")
```
