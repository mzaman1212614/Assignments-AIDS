## Question 2: Electricity Bill Category

```python
units = float(input("Enter the number of units consumed:"))
if units < 100:
    print("Low Usage")
elif units < 300:
    print("Medium Usage")
else:
    print("High Usage")    

```

## Question 3: Online Shopping Discount

```python
total_amount = float(input("Enter total amount:"))
if total_amount < 5000:
    discount_rate = 0
elif total_amount < 9999:
    discount_rate = 0.10
elif total_amount <19999:
    discount_rate = 0.15  
else:
    discount_rate = 0.20
total_discount = total_amount * discount_rate   
final_amount = total_amount - total_discount
print(f'Original amount is: {total_amount}')
print(f'Discount :{total_discount}')
print(f'Final Amount:{final_amount}')

 
```

## Question 4: Employee Salary Grade

```python
salary = int(input("Enter your salary:"))
if salary < 30000:
    print("Grade C")
elif salary <= 59999:
    print("Grade B")    
elif salary <= 99999:
    print("Grade A")   
else:
    print("Grade A+")     
```

## Question 6: Movie Ticket Pricing

```python
age = int(input("Enter your age:"))
if age < 5:
    ticket_price = "Free"
elif age <= 12:
    ticket_price = 300
elif age <= 59:
    ticket_price = 600
else:
    ticket_price = 400
print(f'Ticket Price is {ticket_price}')          
```
